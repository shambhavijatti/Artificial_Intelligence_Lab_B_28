```python
import nltk
import spacy

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import LancasterStemmer
from nltk.stem import SnowballStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

# Download Required Packages
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

# Load spaCy Model
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Natural Language Processing (NLP) is a branch of Artificial Intelligence.
It enables computers to understand human language.
Students are learning NLP using Python.
"""

print(text)
```

```python
nltk.download('punkt_tab')
sentences = sent_tokenize(text)

print("Sentence Tokenization\n")

for i, sentence in enumerate(sentences,1):
    print(i,".",sentence)
```

```python
words = word_tokenize(text)

print(words)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word.lower() not in stop_words]

print("Original Words")

print(words)

print("\nWords after Stop Word Removal")

print(filtered_words)
```

```python
text = """Natural Language Processing (NLP) is a branch of Artificial Intelligence.
It enables computers to understand human language.
Students are learning NLP using Python."""

doc = nlp(text)
```

```python
print("Word Tokenization\n")

for token in doc:
    print(token.text)
```

```python
print("Sentence Tokenization\n")

for sentence in doc.sents:
    print(sentence.text)
```

```python
print("Words after Removing Stop Words\n")

filtered_words = []

for token in doc:
    if not token.is_stop and not token.is_punct:
        filtered_words.append(token.text)

print(filtered_words)
```

```python
print("Lemmatization\n")

for token in doc:
    print(f"{token.text:15} --> {token.lemma_}")
```

```python
porter = PorterStemmer()

print("Porter Stemmer\n")

for word in filtered_words:
    print(word,"-->",porter.stem(word))
```

```python
snow = SnowballStemmer("english")

print("Snowball Stemmer\n")

for word in filtered_words:
    print(word,"-->",snow.stem(word))
```

```python
nltk.download('averaged_perceptron_tagger_eng')
tokens = word_tokenize(text)

pos = pos_tag(tokens)

print("POS Tags\n")

for word,tag in pos:
    print(word," : ",tag)
```

```python
words = ["playing","studies","running","happiness","unhappy"]

lemmatizer = WordNetLemmatizer()

stemmer = PorterStemmer()

print("Word\t\tStem\t\tLemma")

for word in words:
    print(word,"\t",stemmer.stem(word),"\t\t",lemmatizer.lemmatize(word))
```

```python
word = "unhappiness"

prefix = "un"
root = "happy"
suffix = "ness"

print("Word :",word)

print("Prefix :",prefix)

print("Root :",root)

print("Suffix :",suffix)
```

```python
sentence = "I went to the bank."

print(sentence)

print("\nPossible Meaning 1 : Financial Institution")

print("Possible Meaning 2 : River Bank")

print("\nType : Lexical Ambiguity")
```

```python
sentence = "I saw the man with a telescope."

print(sentence)

print("\nInterpretation 1")

print("I used a telescope to see the man.")

print("\nInterpretation 2")

print("The man had a telescope.")

print("\nType : Syntactic Ambiguity")
```

```python
doc = nlp(text)

print("spaCy POS Tags\n")

for token in doc:
    print(f"{token.text:15} {token.pos_:10} {token.tag_:8}")
```

```python
print("Morphological Features\n")

for token in doc:
    print("Word:", token.text)
    print("Lemma:", token.lemma_)
    print("POS:", token.pos_)
    print("Morphology:", token.morph)
    print("-----------------------------")
```

```python
sentence = "I saw the man with a telescope."

doc = nlp(sentence)

print("Sentence:", sentence)

print("\nDependency Parsing")

for token in doc:
    print(token.text, token.dep_, token.head.text)
```
